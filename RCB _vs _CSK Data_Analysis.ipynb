```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
data = {
    "Player": ["Virat Kohli", "Faf du Plessis", "Glenn Maxwell",
               "Rajat Patidar", "Dinesh Karthik",
               "Ruturaj Gaikwad", "MS Dhoni", "Shivam Dube",
               "Ravindra Jadeja", "Devon Conway"],

    "Team": ["RCB", "RCB", "RCB", "RCB", "RCB",
             "CSK", "CSK", "CSK", "CSK", "CSK"],

    "Runs": [741, 730, 400, 333, 140,
             590, 104, 418, 190, 672],

    "Strike_Rate": [154, 153, 183, 152, 140,
                    147, 182, 158, 142, 140]
}

df = pd.DataFrame(data)

df
```

```python
print(df.head())
print(df.shape)
print(df.isnull().sum())
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    y="Player",
    x="Runs",
    hue="Team"
)

plt.title("RCB vs CSK - Player Runs Comparison")
plt.xlabel("Total Runs")
plt.ylabel("Players")

plt.tight_layout()
plt.show()
```

```python
teams = ["RCB", "CSK"]
wins = [14, 21]

plt.figure(figsize=(7, 7))

plt.pie(
    wins,
    labels=teams,
    autopct="%1.1f%%",
    startangle=90,
    explode=(0.1, 0)
)

plt.title("RCB vs CSK - Head to Head Wins")
plt.show()
```

```python
rcb_df = df[df["Team"] == "RCB"]

top_players = rcb_df.sort_values(
    by="Runs",
    ascending=False
).head(5)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=top_players,
    x="Runs",
    y="Player",
    color="red"
)

plt.title("Top 5 RCB Players - Runs Analysis")
plt.xlabel("Total Runs")
plt.ylabel("Player Name")

plt.show()
```

```python
bowling_data = {
    "Player": [
        "Mohammed Siraj", "Harshal Patel",
        "Yuzvendra Chahal", "Glenn Maxwell",
        "Deepak Chahar", "Ravindra Jadeja",
        "Dwayne Bravo", "R Ashwin"
    ],

    "Team": [
        "RCB", "RCB", "RCB", "RCB",
        "CSK", "CSK", "CSK", "CSK"
    ],

    "Wickets": [19, 14, 21, 8, 18, 12, 16, 10]
}

bowl_df = pd.DataFrame(bowling_data)

bowl_df
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=bowl_df,
    x="Wickets",
    y="Player",
    hue="Team"
)

plt.title("RCB vs CSK - Bowling Performance")
plt.xlabel("Total Wickets")
plt.ylabel("Bowlers")

plt.tight_layout()
plt.show()
```

```python
match_data = {
    "Match": [1, 2, 3, 4, 5, 6, 7, 8],
    "RCB_Runs": [172, 185, 165, 198, 210, 175, 190, 205],
    "CSK_Runs": [160, 178, 190, 155, 195, 200, 170, 188]
}

match_df = pd.DataFrame(match_data)

match_df
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    match_df["Match"],
    match_df["RCB_Runs"],
    marker="o",
    label="RCB"
)

plt.plot(
    match_df["Match"],
    match_df["CSK_Runs"],
    marker="o",
    label="CSK"
)

plt.title("RCB vs CSK - Match Wise Runs")
plt.xlabel("Match Number")
plt.ylabel("Total Runs")

plt.legend()
plt.grid()
plt.show()
```
